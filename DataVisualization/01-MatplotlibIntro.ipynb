```python

dataset = pd.read_csv('appl_1980_2014.csv')
dataset.head()
```

```python
X = pd.to_datetime(dataset['Date'])
y = dataset['Adj Close']

# Line Plot
plt.title("DATE VS ADJ CLOSE")
plt.xlabel("DATE")
plt.ylabel("ADJ CLOSE")
plt.plot(X,y,"m")
plt.show()#py file me krrhe ho code
```
