```python
import pandas as pd
import numpy as np
```

```python
data = pd.read_csv("dataset/house.csv")
```

```python
data.head()#first 5 rows
```

```python
data.tail(5)#last 5
```

```python
x = data['livingArea']
y = data['price']
print(len(x))
print(len(y))
```

```python
mean_x = np.mean(x)
mean_y = np.mean(y)
```

```python
num = 0
for i in range(len(x)):
    num += (x[i]-mean_x)*(y[i]-mean_y)

deno = 0
for i in range(len(x)):
    deno += (x[i]-mean_x)**2

B1 = num/deno
```

```python
B1
```

```python
B0 = mean_y - B1*mean_x
```

```python
B0,B1
```

```python
#prediction
newX = 906
Y_pred = B0+B1*newX
```

```python
Y_pred 
```

```python
y[0]
```

```python
error =  y[0]-Y_pred
```
